## Fase 3 — Contrasta cada sistema con la definición legal de «sistema de IA»

El **artículo 3.1 del Reglamento de IA (UE 2024/1689)** define un sistema de IA como un sistema
basado en una máquina, diseñado para funcionar con **distintos niveles de autonomía**, que puede
mostrar **capacidad de adaptación tras el despliegue** y que, para objetivos explícitos o
implícitos, **infiere** de la información de entrada la manera de generar resultados de salida

**La inferencia es la frontera**

El considerando 12 del Reglamento señala que la **capacidad de inferencia** es lo que
distingue un sistema de IA del software convencional, y excluye expresamente los sistemas
basados en reglas definidas solo por personas para ejecutar operaciones automáticamente. Un
formulario con validaciones o una hoja de cálculo con fórmulas **no** son IA, por mucho que se
anuncien como «inteligentes».
