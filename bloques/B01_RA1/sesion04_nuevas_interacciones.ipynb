## Fase 4 — Anticipa las fricciones de la interacción conversacional

La interfaz conversacional **no es automáticamente fácil**. Un estudio de usabilidad del Nielsen
Norman Group (2023, 8 participantes en sesiones de 90 minutos con ChatGPT-3.5) documentó tres
comportamientos recurrentes:

| Fricción | En qué consiste | Qué provoca en el usuario |
|---|---|---|
| *Accordion editing* | El usuario pide una y otra vez ampliar o resumir la respuesta | Iteraciones extra hasta lograr el formato útil |
| *Apple picking* | Debe describir o copiar y pegar trozos de respuestas anteriores porque no puede seleccionarlos | Trabajo manual y pérdida de contexto |
| *Response outlining* | Especifica en el *prompt* la estructura que quiere en la respuesta | Solo lo adopta **después** de recibir una respuesta insatisfactoria |

Responde: ¿cuál de las tres fricciones afectaría más a tu propuesta y **qué cambio concreto de
diseño** la reduciría (botones, plantillas de consulta, respuestas estructuradas, historial
navegable)?
