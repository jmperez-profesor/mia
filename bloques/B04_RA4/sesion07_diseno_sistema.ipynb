## 1. La tabla de selección: la tarea manda

Tarea de la célula LARA: recoger **tarros de miel de 0,5 kg** de una cinta y colocarlos en cajas a **700 mm**, **20 piezas/min**, repetibilidad **±0,1 mm**.

**Payload = pieza + herramienta (EOAT) + cables + sensores.** Con pinza de 0,5 kg: ≈ **1,2 kg**.

````python
modelos = pd.DataFrame({
    "Modelo": ["UR5e", "UR10e", "KUKA KR AGILUS-6", "FANUC M-10iD"],
    "Payload (kg)": [5, 12.5, 6, 10],
    "Alcance (mm)": [850, 1300, 901, 1420],
    "Repetibilidad (mm)": [0.03, 0.05, 0.03, 0.03],
    "Tipo": ["cobot", "cobot", "industrial", "industrial"],
})
modelos["cumple_payload"] = modelos["Payload (kg)"] >= 2.0
modelos["cumple_alcance"] = modelos["Alcance (mm)"] >= 700
modelos["cumple_repetibilidad"] = modelos["Repetibilidad (mm)"] <= 0.1
modelos["candidato"] = modelos[["cumple_payload", "cumple_alcance", "cumple_repetibilidad"]].all(axis=1)
print(modelos.to_string(index=False))
````

## 2. Decisión: ¿colaborativa o vallada?

- Comparte espacio con personas → **aplicación colaborativa** (ISO 10218:2025, limitación de potencia/fuerza, evaluación biomecánica de la tarea completa).
- Célula cerrada de alta cadencia → **vallado con enclavamientos**.
- Ojo: un cobot con una herramienta cortante **no** es una aplicación colaborativa.

````python
# Checklist de la aplicación (ISO 10218:2025)
checklist = {
    "Comparte espacio con personas": True,
    "Herramienta segura en contacto": True,
    "Fuerza/velocidad limitadas y verificadas": True,
    "Evaluacion de riesgos de la tarea completa (ISO 12100)": True,
}
aplicacion_colaborativa = all(checklist.values())
print("Aplicacion colaborativa:", aplicacion_colaborativa)
````

## 3. Esquema de la célula (Industria 4.0)

```mermaid
flowchart LR
    PLC[PLC de seguridad] <-->|PROFINET / EtherCAT| R[Robot]
    R --> S[Sensores: vision, fuerza/par]
    R --> H[EOAT: pinza]
    R -->|OPC UA / MQTT| MES[Plataforma IIoT / MES]
    MES --> DT[Gemelo digital: mantenimiento predictivo]
```

- Bus determinista: **PROFINET / EtherCAT** entre PLC y robot.
- Telemetría: **OPC UA** o **MQTT** al MES.
- **Gemelo digital** con los datos de ciclo y sensores.
