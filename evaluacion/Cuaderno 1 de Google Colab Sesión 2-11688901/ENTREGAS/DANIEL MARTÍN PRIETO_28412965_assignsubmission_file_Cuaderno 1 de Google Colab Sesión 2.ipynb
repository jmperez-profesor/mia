## 1 · La ficha de 1 minuto: las 4 preguntas

Cualquier sistema de IA se puede describir con cuatro preguntas. Son la versión corta del ciclo **percibe → razona → actúa** que has visto en clase.

| # | Pregunta | Qué respondes |
|---|---|---|
| 1 | ¿Qué **PERCIBE**? | Los datos de entrada: texto, imagen, audio, sensores, historial… |
| 2 | ¿Razona con **REGLAS** o con **DATOS**? | *Reglas*: las escribió una persona. *Datos*: las aprendió de ejemplos |
| 3 | ¿Qué **ACCIÓN** produce? | Qué hace al final: responde, recomienda, controla, bloquea… |
| 4 | Tarea **ESTRECHA**: ¿cuál? | **Una sola** tarea y lo que **NO** hace. Toda la IA actual es débil (estrecha) |

**Cómo decidir la casilla 2:** ¿podrías escribir todas las reglas en una hoja? Si sí → **reglas**. Si no, porque hay demasiados casos y cambian → **datos** (el sistema las aprende de ejemplos).

La casilla 4 es la más importante: obliga a decir **qué NO hace** el sistema.

### Las ideas de Python que vas a ver

Esto es todo lo que aparece en el código de este cuaderno:

| Lo que verás | Qué significa |
|---|---|
| `# texto` | **Comentario**: Python lo ignora; es una explicación para personas |
| `"texto"` | Un **texto**. Siempre va entre comillas |
| `nombre = valor` | **Guarda** un valor con un nombre (una *variable*) para usarlo después |
| `{ "clave": "valor", … }` | Un **diccionario**: una lista de pares *nombre → valor*. Una ficha es justo eso |
| `print(…)` | **Escribe** algo en pantalla |
| `if …:` y `for … in …:` | `if` hace algo **solo si** se cumple una condición; `for` **repite** algo una vez por cada elemento de una lista |
| `def nombre(…):` | Crea una **herramienta** reutilizable con nombre propio (una *función*) |

Las dos celdas siguientes crean dos herramientas: **`mostrar_ficha`** (dibuja la ficha) y **`comprobar_ficha`** (revisa los errores más típicos). **Solo hay que ejecutarlas una vez**; no hace falta entender cómo están hechas ni modificarlas. Si no sale nada debajo, es normal.

```python
# ---------------------------------------------------------------------------
# CELDA DE HERRAMIENTAS. Solo hay que ejecutarla (▶); no hace falta modificarla.
# "def" crea una herramienta con nombre propio (una FUNCIÓN) que usaremos después.
# ---------------------------------------------------------------------------

# Herramienta 1: mostrar_ficha. Recibe una ficha y la escribe en pantalla con buen aspecto.
def mostrar_ficha(ficha):
    # ficha.get("casilla", "...") lee lo escrito en una casilla; si no existe, devuelve el valor de la derecha.
    print("=" * 62)                       # "=" * 62 repite el símbolo "=" 62 veces: dibuja una línea
    print(f"FICHA DE 1 MINUTO  -  {ficha.get('sistema', '')}")   # f"..." cambia lo que va entre llaves {} por su valor
    print("=" * 62)
    # Las cuatro líneas siguientes escriben una pregunta de la ficha cada una, con su respuesta al lado.
    print(f"1. ¿Qué PERCIBE?                 {ficha.get('percibe', '')}")
    print(f"2. ¿Razona con REGLAS o DATOS?   {ficha.get('razona_con', '')}")
    print(f"3. ¿Qué ACCIÓN produce?          {ficha.get('accion', '')}")
    print(f"4. Tarea ESTRECHA:               {ficha.get('tarea_estrecha', '')}")
    print("=" * 62)


# Herramienta 2: comprobar_ficha. Revisa la ficha y avisa de los errores más típicos.
def comprobar_ficha(ficha):
    # Lista con los nombres de las casillas que tiene que tener la ficha.
    casillas = ["sistema", "percibe", "razona_con", "accion", "tarea_estrecha"]
    # Lista vacía donde iremos apuntando los problemas que encontremos.
    avisos = []

    # "for" repite lo que hay debajo (con sangría) una vez por cada casilla de la lista.
    for casilla in casillas:
        # Lee lo que has escrito en esa casilla; .strip() quita los espacios sobrantes de los extremos.
        texto = ficha.get(casilla, "").strip()
        # "if" ejecuta lo de debajo solo si la condición es cierta.
        # Aquí: la casilla está vacía o todavía tiene los "..." de relleno.
        if texto == "" or texto.startswith("..."):
            # .append(...) añade un elemento al final de la lista de avisos.
            avisos.append(f"Falta rellenar la casilla '{casilla}'.")

    # La casilla 'razona_con' solo admite dos respuestas: reglas o datos.
    # .lower() pasa el texto a minúsculas para que "Datos" y "datos" cuenten igual.
    if ficha.get("razona_con", "").strip().lower() not in ["reglas", "datos"]:
        avisos.append("'razona_con' debe ser exactamente  reglas  o  datos.")

    # Frases vacías típicas, que no describen nada (por ejemplo, decir solo "usa IA").
    frases_vacias = ["usa ia", "usa inteligencia artificial", "es inteligente", "ayuda a los usuarios"]
    # Se revisan las casillas 1, 3 y 4 (la 2 ya está comprobada arriba).
    for casilla in ["percibe", "accion", "tarea_estrecha"]:
        if ficha.get(casilla, "").strip().lower() in frases_vacias:
            avisos.append(f"La casilla '{casilla}' es demasiado genérica: concreta qué datos entran, qué hace y qué NO hace.")

    # len(...) cuenta cuántos caracteres tiene un texto.
    # Una tarea estrecha de verdad necesita algo de detalle: la tarea y lo que NO hace.
    if len(ficha.get("tarea_estrecha", "").strip()) < 25:
        avisos.append("La 'tarea_estrecha' es muy corta: indica la tarea concreta y lo que NO hace.")

    # len(avisos) == 0 significa que no hemos encontrado ningún problema.
    if len(avisos) == 0:
        print("OK: la ficha está completa y no se detectan errores típicos.")
    else:
        print("Revisa tu ficha:")
        for aviso in avisos:              # escribe cada aviso en una línea
            print("  -", aviso)
```

**✏️ Respuesta**

La IA es más rentable en la **personalización y recomendación**. El catálogo tiene millones de canciones y millones de usuarios, así que sería imposible preparar a mano una lista para cada persona; la IA lo hace cada semana a un coste muy bajo. Además, el beneficio está ligado al negocio: si las recomendaciones aciertan, la gente escucha más y no cancela la suscripción, que es de donde sale la mayor parte de los ingresos.

Luego está la **detección de fraude**, porque protege el dinero que se reparte entre artistas y la confianza en la plataforma, aunque su beneficio se ve menos porque es un sistema interno.

El **DJ con IA** es el menos crítico hoy: aporta valor como novedad y diferencia a Spotify de la competencia, pero el negocio funcionaría sin él. Es una apuesta de futuro más que una fuente principal de rentabilidad.
